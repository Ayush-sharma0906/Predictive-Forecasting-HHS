```python
import pandas as pd
comparison_df = pd.DataFrame({
    "Model": [
        "Baseline",
        "ARIMA",
        "SARIMA",
        "Holt-Winters",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        baseline_mae,
        arima_mae,
        sarima_mae,
        hw_mae,
        rf_mae,
        gb_mae
    ],
    "RMSE": [
        baseline_rmse,
        arima_rmse,
        sarima_rmse,
        hw_rmse,
        rf_rmse,
        gb_rmse
    ],
    "MAPE": [
        baseline_mape,
        arima_mape,
        sarima_mape,
        hw_mape,
        rf_mape,
        gb_mape
    ]
})

comparison_df = comparison_df.sort_values("MAPE")
comparison_df
```

```python
import matplotlib.pyplot as plt

comparison_df.plot(
    x="Model",
    y=["MAE", "RMSE"],
    kind="bar",
    figsize=(10,5),
    rot=20
)

plt.title("Forecasting Model Comparison")
plt.grid(axis="y")
plt.show()
```
